# Part 7 — Automatic Data Analysis

If a CSV was uploaded, this section creates:

- shape;
- missing-value analysis;
- numeric descriptive statistics;
- duplicate count;
- candidate timestamp detection;
- candidate categorical distributions.

This is intended as **supporting analysis**, not as a substitute for project-specific engineering interpretation.


````python
analysis_summary = {
    "analysis_file": None,
    "rows": 0,
    "columns": 0,
    "duplicate_rows": 0,
    "missing_cells": 0,
    "numeric_columns": [],
    "categorical_columns": []
}

numeric_summary = pd.DataFrame()
missing_summary = pd.DataFrame()
category_summaries = {}

if analysis_df is not None:

    analysis_summary["analysis_file"] = ANALYSIS_FILE.name
    analysis_summary["rows"] = int(len(analysis_df))
    analysis_summary["columns"] = int(len(analysis_df.columns))
    analysis_summary["duplicate_rows"] = int(
        analysis_df.duplicated().sum()
    )
    analysis_summary["missing_cells"] = int(
        analysis_df.isna().sum().sum()
    )

    numeric_cols = (
        analysis_df
        .select_dtypes(include="number")
        .columns
        .tolist()
    )

    categorical_cols = [
        c
        for c in analysis_df.columns
        if c not in numeric_cols
        and analysis_df[c].nunique(dropna=True) <= 20
    ]

    analysis_summary["numeric_columns"] = numeric_cols
    analysis_summary["categorical_columns"] = categorical_cols

    missing_summary = (
        analysis_df
        .isna()
        .sum()
        .rename("missing_count")
        .reset_index()
        .rename(columns={"index": "column"})
    )

    numeric_summary = (
        analysis_df[numeric_cols]
        .describe()
        .T
        .reset_index()
        .rename(columns={"index": "column"})
        if numeric_cols
        else pd.DataFrame()
    )

    for col in categorical_cols[:6]:
        category_summaries[col] = (
            analysis_df[col]
            .value_counts(dropna=False)
            .head(20)
            .rename_axis(col)
            .reset_index(name="count")
        )

    print(json.dumps(analysis_summary, indent=2))

    print("\nMissing values:")
    display(missing_summary)

    if not numeric_summary.empty:
        print("\nNumeric descriptive statistics:")
        display(numeric_summary)

    for col, table in category_summaries.items():
        print(f"\nDistribution: {col}")
        display(table)

else:
    print("No CSV analysis was requested.")
````

# Part 13 — Install and Run Ollama

The capstone notebook uses **Ollama + qwen2.5:0.5b** for four review tasks:

1. Capstone completeness audit
2. Engineering evidence review
3. Documentation/handoff review
4. Red-team review

AI may assist development and reporting, but the student remains responsible for verifying the final work.


````python
import subprocess
import shutil as _shutil
import time as _time
import requests

def shell(cmd):
    result = subprocess.run(
        cmd,
        shell=True,
        text=True,
        capture_output=True
    )

    if result.stdout:
        print(result.stdout[-2500:])

    if result.returncode != 0 and result.stderr:
        print(result.stderr[-2500:])

    return result.returncode

shell("apt-get update -qq")
shell("apt-get install -y -qq curl zstd")

if _shutil.which("ollama") is None:
    shell(
        "curl -fsSL https://ollama.com/install.sh | sh"
    )

OLLAMA_URL = "http://127.0.0.1:11434"
OLLAMA_MODEL = "qwen2.5:0.5b"

def ollama_running():
    try:
        return (
            requests.get(
                f"{OLLAMA_URL}/api/tags",
                timeout=3
            ).status_code == 200
        )
    except requests.RequestException:
        return False

if not ollama_running():

    ollama_log = open(
        "/tmp/ollama_module8_capstone.log",
        "a"
    )

    ollama_process = subprocess.Popen(
        ["ollama", "serve"],
        stdout=ollama_log,
        stderr=subprocess.STDOUT
    )

    for _ in range(45):
        if ollama_running():
            break
        _time.sleep(1)

if not ollama_running():
    raise RuntimeError(
        "Ollama did not start successfully."
    )

shell(
    f"ollama pull {OLLAMA_MODEL}"
)

print("✅ Ollama ready:", OLLAMA_MODEL)
````

````python
AI_SYSTEM_POLICY = """
You are an engineering-software capstone review assistant for EGN 321.

Use only evidence supplied by the student.

Rules:
1. Do not invent test results, files, calculations, readings, logs, failures, timestamps, or application features.
2. Separate observation, inference, hypothesis, recommendation, and limitation.
3. Deterministic tests, source data, logs, calculations, and observed application behavior are authoritative.
4. If evidence is insufficient, identify exactly what is missing.
5. Never claim a root cause is proven unless deterministic evidence proves it.
6. Challenge vague claims such as 'it works' or 'the AI verified it.'
7. Prefer recommendations that can be verified using a deterministic test or observable success condition.
8. Keep the response suitable for inclusion in a professional student report.
"""

def ask_ollama(
    prompt,
    timeout=600,
    num_predict=1000
):
    payload = {
        "model": OLLAMA_MODEL,
        "prompt": prompt,
        "system": AI_SYSTEM_POLICY,
        "stream": False,
        "keep_alive": "30m",
        "options": {
            "temperature": 0.2,
            "num_predict": num_predict,
            "num_ctx": 4096
        }
    }

    response = requests.post(
        f"{OLLAMA_URL}/api/generate",
        json=payload,
        timeout=(10, timeout)
    )

    response.raise_for_status()

    return response.json()["response"].strip()

print(
    "Warm-up:",
    ask_ollama(
        "Reply with exactly READY.",
        num_predict=10
    )
)
````

# Part 24 — Create the Final Named Submission ZIP

The student's name and project title are automatically added to the ZIP filename.


````python
# ============================================================
# PART 24 — CREATE FINAL NAMED SUBMISSION ZIP
# ============================================================

from pathlib import Path
import zipfile

# ------------------------------------------------------------
# Verify required variables exist
# ------------------------------------------------------------

required_variables = [
    "STUDENT_FILE_NAME",
    "PROJECT_FILE_NAME",
    "STUDENT_NAME",
    "PROJECT_TITLE",
    "EXPORT_DIR"
]

missing_variables = [
    name
    for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise RuntimeError(
        "The following required variables are missing: "
        + ", ".join(missing_variables)
        + "\n\nPlease rerun Part 0 and the previous export sections before Part 24."
    )

# ------------------------------------------------------------
# Verify export directory
# ------------------------------------------------------------

EXPORT_DIR = Path(EXPORT_DIR)

if not EXPORT_DIR.exists():
    raise FileNotFoundError(
        f"Export directory does not exist:\n{EXPORT_DIR}\n\n"
        "Run the previous report/export steps first."
    )

# ------------------------------------------------------------
# Create final ZIP filename
# ------------------------------------------------------------

ZIP_FILE = Path(
    f"/content/"
    f"{STUDENT_FILE_NAME}_"
    f"{PROJECT_FILE_NAME}_"
    f"EGN321_Capstone_Submission.zip"
)

# Remove old ZIP if it already exists
if ZIP_FILE.exists():
    ZIP_FILE.unlink()

# ------------------------------------------------------------
# Find files to include
# ------------------------------------------------------------

files_to_package = [
    p
    for p in EXPORT_DIR.rglob("*")
    if p.is_file()
]

if not files_to_package:
    raise RuntimeError(
        f"No files were found inside:\n{EXPORT_DIR}\n\n"
        "Run the previous export/report sections before creating the ZIP."
    )

# ------------------------------------------------------------
# Build ZIP
# ------------------------------------------------------------

with zipfile.ZipFile(
    ZIP_FILE,
    "w",
    zipfile.ZIP_DEFLATED
) as z:

    for p in files_to_package:

        archive_name = str(
            p.relative_to(
                EXPORT_DIR
            )
        )

        z.write(
            p,
            arcname=archive_name
        )

# ------------------------------------------------------------
# Verify ZIP
# ------------------------------------------------------------

if not ZIP_FILE.exists():
    raise RuntimeError(
        "ZIP creation failed."
    )

with zipfile.ZipFile(
    ZIP_FILE,
    "r"
) as z:

    zip_contents = z.namelist()

    bad_files = z.testzip()

# ------------------------------------------------------------
# Final summary
# ------------------------------------------------------------

print("=" * 70)
print("✅ FINAL CAPSTONE PACKAGE CREATED")
print("=" * 70)

print("Student:", STUDENT_NAME)
print("Project:", PROJECT_TITLE)

print("\nZIP:")
print(ZIP_FILE.name)

print("\nLocation:")
print(ZIP_FILE)

print("\nFiles included:")
for name in zip_contents:
    print(" -", name)

print("\nTotal files:")
print(len(zip_contents))

print("\nZIP size:")
print(
    f"{ZIP_FILE.stat().st_size / 1024 / 1024:.2f} MB"
)

if bad_files is None:
    print("\n✅ ZIP integrity check passed.")
else:
    print(
        "\n⚠️ ZIP integrity problem detected with:",
        bad_files
    )

print("\n" + "=" * 70)
print("CAPSTONE PACKAGE READY")
print("=" * 70)

# ------------------------------------------------------------
# OPTIONAL — Automatically download
# ------------------------------------------------------------

# Uncomment these two lines if you want the ZIP
# to download automatically after creation.

# from google.colab import files
# files.download(str(ZIP_FILE))
````
